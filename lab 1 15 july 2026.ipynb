```python
room = ["D", "D", "C", "D", "C", "D", "D", "C", "D"]
def show_room(room):
  picture = ""
  for spot in room:
    if spot == "D":
      picture += " 💩 "
    else:
      picture += " ✨ "
  print(picture)
show_room(room)
```

```python
counter = 0
```

```python
def clean_spot(spot):
  if spot == "D":
    spot = "C"
    return "C"
  else:
    return "C"
result = clean_spot("D")
print("the robot looked at dirty spot 💩 and made it:", result)
```

```python
print("Before - the dirty room")
show_room(room)
print()

for i in range(len(room)):
  if room[i] == "D":
    counter += 1
  room[i] = clean_spot(room[i])
  print("after cleaning the spot", i+1, ":")
  show_room(room)
  
print()
print("After - all done!")
show_room(room)
print("the robot cleaned", counter, "dirty spots 💩 and made them clean ✨")
```

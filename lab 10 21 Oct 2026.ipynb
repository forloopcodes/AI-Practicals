```python
transitions = {
    'A': {'Right': 'B', 'Down': 'C'},
    'B': {'Left': 'A', 'Down': 'G'},
    'C': {'Right': 'G', 'Up': 'A'},
    'G': {}
}
rewards = {
    'A': -1,
    'B': -1,
    'C': -1,
    'G': 10
}
gamma = 0.9
```

```python
values = {
  'A':0,
  'B':0,
  'C':0,
  'G':0
}
print(values)
```

```python
theta = 1e-4
non_terminal_states = [s for s in transitions if transitions[s]]
while True:
    delta = 0
    new_values = values.copy()
    for state in non_terminal_states:
        q_values = []
        for action, next_state in transitions[state].items():
            future_val = 0 if next_state == 'G' else values[next_state]
            q_val = rewards[next_state] + gamma * future_val
            q_values.append(q_val)
        new_values[state] = max(q_values)
        delta = max(delta, abs(new_values[state] - values[state]))
    values = new_values
    if delta < theta:
        break
final_values = {s: values[s] for s in non_terminal_states}
print("Final Values:", final_values)
```

```python
policy = {}
for state, actions in transitions.items():
    if not actions:
        continue
    best_action = None
    best_q = float('-inf')
    for action, next_state in actions.items():
        future_val = 0 if next_state == 'G' else values[next_state]
        q_val = rewards[next_state] + gamma * future_val
        if q_val > best_q:
            best_q = q_val
            best_action = action
    policy[state] = best_action
print("Optimal Policy:", policy)
```

```python
for state, action in policy.items():
    print(f"From state {state}, take action '{action}'.")
```

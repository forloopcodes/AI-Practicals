```python
def laptop_expert(turns_on , internet_works , is_slow):
    if turns_on =="no":
        return "Check the power cable "
    
    elif internet_works=="no":
        return "Check the wifi and restart the router"
    
    elif is_slow == "yes":
        return "Closed the unused program and restart the system "
    
    else:
        return "N computer problem"
```

```python
turns_on       = input("Enter yes/no if your device turns on or not : ").lower()
internet_works = input("Enter yes/no , if your device is connected to the internet or not : ").lower()
is_slow        = input("Enter yes/no , if your device works slow or laggy : ").lower()


advice = laptop_expert(turns_on , internet_works , is_slow)
print("\n Expert System Adivce : ")
print(advice)
```

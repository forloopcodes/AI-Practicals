```python
# The map: each room lists the rooms you can walk to next.
map_rooms = {
    "A": ["B", "C"],   # from A you can go to B or C
    "B": ["D"],        # from B you can go to D
    "C": ["D"],        # from C you can go to D
    "D": ["G"],        # from D you can go to G
    "G": [],           # G is the treasure! No need to go further.
}

print("Our map:")
print("            A  (Start)")
print("           / \\")
print("          B   C")
print("           \\ /")
print("            D")
print("            -")
print("            G  (Treasure 💎)")
print()
print("Doors from each room:", map_rooms)
```

```python
def bfs(start, goal):
  to_do = [start]  # queue of rooms to explore
  visited = [] # rooms which have already been explored
  order = []  # order of rooms visited
  
  while to_do:
    room = to_do.pop(0)  # get the next room to explore
    if room in visited:
      continue  # skip if already visited
    visited.append(room)  # mark it as visited
    order.append(room)  # add it to the order of rooms visited
    if room == goal:
      return order 
    for neighbor in map_rooms[room]:
      to_do.append(neighbor)  # add neighboring rooms to explore next
  return order  # return the order of rooms visited if goal not found
        
print("Starting BFS from A to G...")
path = bfs("A", "G")
if path:
    print("Path to treasure found:", path)
    print("Length of path:", len(path))
```

```python
def dfs(start, goal):
  to_do = [start]  # queue of rooms to explore
  visited = [] # rooms which have already been explored
  order = []  # order of rooms visited
  
  while to_do:
    room = to_do.pop()  # get the next room to explore
    if room in visited:
      continue  # skip if already visited
    visited.append(room)  # mark it as visited
    order.append(room)  # add it to the order of rooms visited
    if room == goal:
      return order 
    for neighbor in map_rooms[room]:
      to_do.append(neighbor)  # add neighboring rooms to explore next
  return order  # return the order of rooms visited if goal not found
        
print("Starting DFS from A to G...")
path = dfs("A", "G")
if path:
    print("Path to treasure found:", path)
    print("Length of path:", len(path))
```

```python
door_cost = {
  ("A", "B"): 1,
  ("B", "D"): 1,
  ("A", "C"): 5,
  ("C", "D"): 1,
  ("D", "G"): 1
}
def path_cost(path):
  total_cost = 0
  for i in range(len(path) - 1):
    door = (path[i], path[i + 1])
    total_cost += door_cost.get(door, 0)  # get the cost of the door, default to 0 if not found
  return total_cost

path1 = ["A", "B", "D", "G"]
path2 = ["A", "C", "D", "G"]

print("Cost of path", path1, "is", path_cost(path1))
print("Cost of path", path2, "is", path_cost(path2))
print("The cheapest path is", path1 if path_cost(path1) < path_cost(path2) else path2)
```

```python
door_cost[("A", "C")] = 40
door_cost[("D", "G")] = 20
```

```python
cost1 = path_cost(path1)
cost2 = path_cost(path2)

# ✏️ YOUR TURN: we want the cheaper path. Replace HERE with the < sign.
if cost1 < cost2:
    best_path = path1
    best_cost = cost1
else:
    best_path = path2
    best_cost = cost2

print("Path 1 cost:", cost1)
print("Path 2 cost:", cost2)
print()
print("The cheapest path is:", best_path, "with cost", best_cost, "💰")

bfs_order = bfs("A", "G")
dfs_order = dfs("A", "G")

print("RESULTS")
print("-" * 40)
print("BFS  (nearest first) checked:", len(bfs_order), "rooms ->", bfs_order)
print("DFS  (deep first)    checked:", len(dfs_order), "rooms ->", dfs_order)
print("Cheapest path:", best_path, "with cost", best_cost)
print("-" * 40)
print()
print("What we learned:")
print(" * BFS checks the nearest rooms first (spreads out).")
print(" * DFS dives deep down one path first.")
print(" * Cheapest path counts door costs and picks the lowest total.")
```
